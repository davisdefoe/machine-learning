```python
import csv

# Read data from CSV file
with open("training_data.csv", "r") as file:
    reader = csv.reader(file)
    data = list(reader)

# Remove header
header = data[0]
data = data[1:]

# Number of attributes
num_attributes = len(header) - 1

# Most specific hypothesis
S = ["Ø"] * num_attributes

# Most general hypothesis
G = ["?"] * num_attributes


# Check whether hypothesis covers example
def covers(h, example):
    for i in range(num_attributes):
        if h[i] != "?" and h[i] != example[i]:
            return False
    return True


# Candidate Elimination
for example in data:

    attributes = example[:-1]
    label = example[-1]

    # Positive example
    if label == "Yes":

        # Generalize S
        for i in range(num_attributes):

            if S[i] == "Ø":
                S[i] = attributes[i]

            elif S[i] != attributes[i]:
                S[i] = "?"

        # Remove G hypotheses that don't cover positive example
        if not covers(G, attributes):
            G = S.copy()

    # Negative example
    else:

        # Specialize G
        for i in range(num_attributes):

            if G[i] == "?":

                # Use S value to specialize
                if S[i] != "Ø" and S[i] != attributes[i]:
                    G[i] = S[i]

# Display results
print("Specific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
print(G)

print("\nSet of Consistent Hypotheses:")
print(S)
```
